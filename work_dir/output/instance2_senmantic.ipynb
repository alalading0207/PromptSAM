## instances--->secgmention results

2022 dataset

```python
import os
import cv2
import numpy as np

id = "20260123_114157_220"
instance_dir = f"/work_dirs/output/{id}/vis"
semantic_dir = f"/work_dirs/output/{id}/semantic"
os.makedirs(semantic_dir, exist_ok=True)

for name in os.listdir(instance_dir):
    if not name.endswith(".tif"):
        continue


    img = cv2.imread(os.path.join(instance_dir, name), cv2.IMREAD_UNCHANGED)
    img = np.max(img, axis=2)   
    img_binary = np.zeros_like(img, dtype=np.uint8)
    img_binary[img > 0] = 255
    cv2.imwrite(os.path.join(semantic_dir, name), img_binary)
```
